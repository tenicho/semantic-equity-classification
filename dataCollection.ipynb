# Data Collection

Pulls source data from the MySQL database into pandas DataFrames for downstream processing.

````python
import os

import pandas as pd
import mysql.connector
from dotenv import load_dotenv
````

## Database connection

Credentials are read from environment variables (via a local `.env` file, not committed to git) rather than hardcoded here. Create a `.env` file in the project root with:

```
DB_HOST=your-db-host
DB_USER=your-username
DB_PASSWORD=your-password
DB_NAME=your-database-name
```

````python
load_dotenv()

DB_HOST = os.environ["DB_HOST"]
DB_USER = os.environ["DB_USER"]
DB_PASSWORD = os.environ["DB_PASSWORD"]
DB_NAME = os.environ["DB_NAME"]

conn = mysql.connector.connect(
    host=DB_HOST,
    user=DB_USER,
    password=DB_PASSWORD,
    database=DB_NAME,
)
````

## Pull data

Example query — replace with the actual query/table you need.

````python
query = "SELECT ticker, companyName, description FROM PROD.mainStats"
df = pd.read_sql(query, conn)
df.head()
````

````python
os.makedirs("data", exist_ok=True)
df.to_parquet("data/companyStatesSlim.parquet", index=False)
````
